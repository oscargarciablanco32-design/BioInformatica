#### 6.- Realiza una función que realice una digestión con las siguientes enzimas de restricción en nuestra secuencia. La función tiene que devolver la siguiente información: (2 pto)
1. Enzimas que tienen sitios de corte en la secuencia y posición (o posiciones de corte)
2. Enzimas que no tienen sitios de corte (si las hay)
3. Resultado de la digestión: los fragmentos resultantes de nuestra secuencia
4. Indicar número y tamaño de cada fragmento

```python
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI' : 'G*GATCC',
    'AluI' : 'AG*CT',
    'Sau3AI' : '*GATC',
    'EcoRI' : 'G*AATTC'
}
```

Esta **función de digestion** devuelve:
1. enzimas que **si** cortan y sus posiciones de corte
2. enzimas que **no** cortan
3. los fragmentos resultantes
4. número y tamaño de cada fragmento

```python
import re
from typing import Dict, List

# Diccionario con enzimas y su sitio de corte.
# Convención: el asterisco (*) marca el punto exacto de corte dentro del sitio.
enzR = {
    'HindIII' : 'A*AGCTT',
    'BamHI'   : 'G*GATCC',
    'AluI'    : 'AG*CT',
    'Sau3AI'  : '*GATC',
    'EcoRI'   : 'G*AATTC'
}

# Helpers (funciones auxiliares)

def _parse_pattern(pat: str):
    """
    Recibe el patrón con asterisco (p.ej. 'A*AGCTT') y devuelve:
      - site: el motivo sin asterisco (p.ej. 'AAGCTT')
      - cut_off: desplazamiento del corte respecto al INICIO del motivo (0-based).
                 En 'A*AGCTT' el corte es después de la primera 'A' -> offset = 1.
    Esto nos permite calcular 'posición de corte = inicio_del_sitio + cut_off'.
    """
    cut_off = pat.index('*')          # posición del asterisco dentro del motivo
    site = pat.replace('*', '')       # motivo limpio para buscarlo en la secuencia
    return site, cut_off

def _find_starts_overlapping(seq: str, site: str) -> List[int]:
    """
    Busca TODAS las apariciones (incluidas superpuestas) del motivo 'site' en 'seq'.
    Usamos una lookahead '(?=...)' para que re.finditer no consuma caracteres y
    permita coincidencias solapadas. Devolvemos los índices de inicio (0-based).
    """
    return [m.start() for m in re.finditer(f'(?={re.escape(site)})', seq)]

# Núcleo: digestión

def digestion(secuencia: str, enzymes: Dict[str, str]):
    """
    Realiza la digestión simultánea de 'secuencia' (lineal) con todas las 'enzymes'.
    Devuelve un diccionario con:
      - enzymes_with_sites: info detallada de las enzimas que sí cortan
      - enzymes_without_sites: enzimas que no encuentran su sitio
      - fragments: lista de fragmentos resultantes (en orden 5'→3')
      - fragment_sizes: longitud de cada fragmento
      - n_fragments: cantidad total de fragmentos
    """
    seq = secuencia.upper()           # normalizamos para que coincida con los motivos
    L = len(seq)

    enzymes_with_sites = {}
    enzymes_without_sites = []
    all_cuts = []                     # acumularemos TODAS las posiciones de corte (0..L)

    for name, pat in enzymes.items():
        site, cut_off = _parse_pattern(pat)       # quitamos '*' y obtenemos offset de corte
        starts = _find_starts_overlapping(seq, site)  # todos los inicios del sitio
        if starts:
            # Convertimos inicios de sitio → posiciones de corte absolutas en la secuencia
            cuts = [s + cut_off for s in starts]
            enzymes_with_sites[name] = {
                'site': site,
                'cut_offset': cut_off,
                'start_indices': starts,     # dónde empieza cada sitio
                'cut_positions': cuts        # dónde se realiza el corte (0-based)
            }
            all_cuts.extend(cuts)            # agregamos para construir los fragmentos globales
        else:
            enzymes_without_sites.append(name)

    # Para generar fragmentos:
    #  - Ordenamos y desduplicamos puntos de corte
    #  - Incluimos los extremos 0 (inicio) y L (final) para formar todos los tramos
    cut_points = sorted(set([0] + all_cuts + [L]))

    # Fragmentos = rebanadas entre puntos de corte consecutivos
    fragments = [seq[cut_points[i]:cut_points[i+1]] for i in range(len(cut_points)-1)]
    fragment_sizes = [len(f) for f in fragments]

    return {
        'enzymes_with_sites': enzymes_with_sites,
        'enzymes_without_sites': enzymes_without_sites,
        'fragments': fragments,
        'fragment_sizes': fragment_sizes,
        'n_fragments': len(fragments)
    }

def pretty_report(result, preview_nt: int = 10):
    """
    Imprime un resumen compacto:
      - Enzimas que cortan + posiciones
      - Enzimas que no cortan
      - Lista de fragmentos con longitud y extremos (head/tail)
    'preview_nt' controla cuántos nucleótidos mostrar de cada extremo.
    """
    print("1) Enzimas con sitio y posiciones de corte:")
    if result['enzymes_with_sites']:
        for enz, info in result['enzymes_with_sites'].items():
            print(f"   - {enz}: sitio {info['site']} (corte offset {info['cut_offset']})")
            print(f"     inicios del sitio: {info['start_indices']}")
            print(f"     posiciones de corte: {info['cut_positions']}")
    else:
        print("   (ninguna)")

    if result['enzymes_without_sites']:
        print("\n2) Enzimas sin sitio de corte:")
        print("   " + ", ".join(result['enzymes_without_sites']))
    else:
        print("\n2) Todas las enzimas tienen al menos un sitio de corte.")

    print("\n3) Fragmentos resultantes (solo tamaño y extremos):")
    for i, frag in enumerate(result['fragments'], 1):
        if len(frag) > preview_nt * 2:
            # Construimos una vista previa: head...tail para evitar salidas enormes
            muestra = f"{frag[:preview_nt]}...{frag[-preview_nt:]}"
        else:
            muestra = frag
        print(f"   F{i:02d} | len={len(frag)} | {muestra}")

    print(f"\n4) Número de fragmentos: {result['n_fragments']}")
    print("   Tamaños:", result['fragment_sizes'])

    # He reducido la informacion devuelta a los 10 primeros nucleotidos ... 10 últimos nucleotidos para evitar problemas con la memoria y flujo de procesamiento de datos

# EJECUCIÓN:
resultado = digestion(secuencia, enzR)
pretty_report(resultado)  # puedes ajustar preview_nt si quieres ver más/menos extremos
```

Resúmen de la lógica aplicada:

1. Modelo del sitio de corte: cada motivo lleva un * para marcar el offset del corte dentro del sitio (0-based). Lo quito para buscar, pero guardo el offset para calcular la posición real de corte.

2. Búsqueda de sitios solapados: uso re.finditer con una lookahead (?=motivo) para capturar coincidencias superpuestas (importante con motivos como GATC).

3. Cortes globales: convierto cada inicio de motivo s en corte absoluto s + offset. Junto todos los cortes de todas las enzimas.

4. Construcción de fragmentos: ordeno y desduplico los cortes, añado extremos 0 y L, y genero fragmentos como rebanadas entre puntos adyacentes.

5. Salida segura para Jupyter: el reporte muestra tamaños y una vista previa (head...tail) para evitar el error “IOPub data rate exceeded” cuando la secuencia es larga. Si necesitas todo el texto, puedes guardarlo a archivo en vez de imprimirlo.

#### 7.- Realiza una función que mediante la introducción de una secuencia de ADN devuelva lo siguiente: (3pto)
1. Todos los ORFs posibles que puedan surgir de la secuencia.
2. Selecciona los 6 ORFs más largos e indica si corresponden a la strand positiva o negativa.
3. Devuelve como mejor predicción, el más largo de todos.
4. Calcula el porcentaje de aminoácidos aromáticos de esa proteína elegida y su longitud relativa a la secuencia original de ADN.

```python
from typing import List, Dict, Tuple   # Importamos tipos opcionales para anotar funciones

# Funciones auxiliares

def reverse_complement(seq: str) -> str:
    """Devuelve el reverso complementario de una secuencia de ADN."""
    complement = str.maketrans("ACGTacgt", "TGCAtgca")  # tabla para cambiar A↔T y C↔G
    return seq.translate(complement)[::-1]              # traducimos y luego invertimos

def translate_codon(codon: str) -> str:
    """Traduce un codón de ADN en su aminoácido de 1 letra."""
    codon = codon.upper()   # nos aseguramos que está en mayúsculas
    table = {              # tabla de traducción estándar (simplificada)
        'TAA': '*','TAG': '*','TGA': '*',   # codones STOP
        'TTT':'F','TTC':'F','TTA':'L','TTG':'L',
        'CTT':'L','CTC':'L','CTA':'L','CTG':'L',
        'ATT':'I','ATC':'I','ATA':'I','ATG':'M',  # ATG = Met (inicio)
        'GTT':'V','GTC':'V','GTA':'V','GTG':'V',
        'TCT':'S','TCC':'S','TCA':'S','TCG':'S',
        'CCT':'P','CCC':'P','CCA':'P','CCG':'P',
        'ACT':'T','ACC':'T','ACA':'T','ACG':'T',
        'GCT':'A','GCC':'A','GCA':'A','GCG':'A',
        'TAT':'Y','TAC':'Y','CAT':'H','CAC':'H',
        'CAA':'Q','CAG':'Q','AAT':'N','AAC':'N',
        'AAA':'K','AAG':'K','GAT':'D','GAC':'D',
        'GAA':'E','GAG':'E','TGT':'C','TGC':'C',
        'TGG':'W','CGT':'R','CGC':'R','CGA':'R','CGG':'R',
        'AGT':'S','AGC':'S','AGA':'R','AGG':'R',
        'GGT':'G','GGC':'G','GGA':'G','GGG':'G'
    }
    return table.get(codon, "")   # devolvemos el aa o "" si el codón no existe

# 1) Buscar todos los ORFs

def find_orfs(seq: str) -> List[Dict]:
    """
    Busca todos los ORFs (Open Reading Frames) en ambas hebras y los 3 marcos.
    Devuelve una lista de diccionarios con info de cada ORF.
    """
    seq = seq.upper()                     # normalizamos a mayúsculas
    rev = reverse_complement(seq)         # calculamos la hebra inversa complementaria
    stops = {"TAA", "TAG", "TGA"}         # codones de STOP
    starts = {"ATG"}                      # codón de inicio (puedes ampliar si quieres)
    orfs = []                             # aquí guardaremos todos los ORFs

    # Función interna para escanear una hebra concreta
    def scan(strand_seq, strand_sign):
        for frame in range(3):                      # recorremos los 3 marcos posibles 
            i = frame                              # posición inicial del marco
            while i < len(strand_seq) - 2:         # mientras queden codones completos
                codon = strand_seq[i:i+3]          # leemos el codón actual
                if codon in starts:                # si es ATG, posible inicio de ORF
                    j = i                          # apuntamos donde empezó
                    aa_seq = []                    # almacenaremos los aa traducidos
                    while j < len(strand_seq) - 2:
                        cod = strand_seq[j:j+3]    # codón en curso
                        if cod in stops:           # si encontramos STOP
                            nt_seq = strand_seq[i:j+3]  # nucleótidos desde ATG hasta STOP
                            orfs.append({          # guardamos un diccionario con los datos
                                "strand": strand_sign,           # hebra (+ o -), donde la hebra lider es la + y la rezagada la -
                                "frame": frame,                 # marco (0,1,2), (cuando traducimos el ADN a proteinas lo hacemos usando codones o tripletes, como cada hebra puede leerse empezando en tres posiciones distintas, hay tres marcos posibles en cada dirección, donde el primer nucleotido es frame = 0 el segundo frame = 1 y el tercero frame = 2. Combinando las dos hebras y los tres frames por hebra tenemos 6 marcos de lectura en total, por eso se suele hablar de “los seis ORFs”)
                                "start": i if strand_sign == '+' else len(seq)-j-3,
                                "end": j+3 if strand_sign == '+' else len(seq)-i,
                                "nt_seq": nt_seq,               # secuencia nucleotídica
                                "aa_seq": ''.join(aa_seq)       # proteína sin '*'
                            })
                            break                  # salimos de este ORF
                        aa = translate_codon(cod)  # traducimos codón
                        if aa:                     # si es válido
                            aa_seq.append(aa)      # lo añadimos
                        else:
                            break                  # codón inválido → terminamos
                        j += 3                     # avanzamos al siguiente codón
                i += 3                            # avanzamos al siguiente posible inicio

    scan(seq, '+')    # exploramos la hebra directa
    scan(rev, '-')    # exploramos la hebra complementaria
    return orfs       # devolvemos la lista de ORFs encontrados

# 2) Elegir los 6 ORFs más largos y 3) el mejor

def select_top_orfs(orfs: List[Dict], top_n: int = 6) -> Tuple[List[Dict], Dict]:
    """
    Ordena los ORFs por tamaño (longitud de proteína) y devuelve:
      - los top_n más largos
      - el ORF más largo (posición 0)
    """
    sorted_orfs = sorted(orfs, key=lambda o: len(o['aa_seq']), reverse=True)
    return sorted_orfs[:top_n], sorted_orfs[0] if sorted_orfs else None

# 4) % de aromáticos y longitud relativa

def aromatic_percentage(protein: str) -> float:
    """Calcula el porcentaje de aa aromáticos (F, Y, W)."""
    if not protein:
        return 0.0
    return 100 * sum(1 for aa in protein if aa in {'F','Y','W'}) / len(protein)

def relative_length(nt_orf: str, original_seq: str) -> float:
    """Calcula longitud del ORF respecto al tamaño de la secuencia original."""
    if not original_seq:
        return 0.0
    return 100 * len(nt_orf) / len(original_seq)

# Función principal que une todo

def analyze_orfs(dna_seq: str):
    dna_seq = dna_seq.upper()           # nos aseguramos de trabajar en mayúsculas
    all_orfs = find_orfs(dna_seq)       # buscamos todos los ORFs

    if not all_orfs:                    # si la lista está vacía
        print("No se encontraron ORFs.")# mensaje informativo
        return

    top6, best = select_top_orfs(all_orfs, 6)  # elegimos los 6 mejores y el top

    # Imprimimos resultados
    print(f"Se encontraron {len(all_orfs)} ORFs.\n")
    print("== Los 6 ORFs más largos ==")
    for i, o in enumerate(top6, 1):
        print(f"{i}. {len(o['aa_seq'])} aa | strand {o['strand']} | frame {o['frame']}")

    print("\n== Mejor predicción (más largo) ==")
    print(f"Longitud (aa): {len(best['aa_seq'])}")
    print(f"Strand: {best['strand']}, frame: {best['frame']}")

    # calculamos estadísticas de la proteína elegida
    arom = aromatic_percentage(best['aa_seq'])
    rel = relative_length(best['nt_seq'], dna_seq)
    print(f"\nPorcentaje de aa aromáticos: {arom:.2f}%")
    print(f"Longitud relativa del ORF: {rel:.2f}% del ADN original")

    # devolvemos todo en un dict por si se quiere seguir trabajando
    return {
        "all_orfs": all_orfs,
        "top6": top6,
        "best": best,
        "aromatic_pct": arom,
        "relative_length_pct": rel
    }

# EJECUCIÓN usando tu variable `secuencia`

resultado_orfs = analyze_orfs(secuencia)  # llamamos a la función con tu ADN
```
