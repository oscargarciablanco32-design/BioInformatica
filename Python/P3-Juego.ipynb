## JUEGO DE ADIVINANZAS

#### Escribe un programa que coja un número entero del 1 al 100, y los jugadores tienen que tratar de adivinar el número.

#### Las reglas son las siguientes:

1. Si un jugador escoje un número menor que 1 o mayor que 100, el programa dice "FUERA DE LÍMITES"
2. En el primer turno del jugador, si está en un rango de 10 del número dice "¡CALIENTE!". Si está fuera de ese rango dice "¡FRIO!"
3. En todas las rondas siguientes, si el jugador se acerca más que la vez anterior, dice "¡MAS CALIENTE!". Si está más lejos que la anterior vez dice "¡MAS FRIO!"
4. Cuando el jugador acierte el número, les dirá que han acertado y cuántos intentos les ha llevado





##### PISTAS
1. Trata obtener un número aleatorio entre el 1 y el 100
2. Haz una introducción de tu juego y explica las reglas
3. Crea una lista que almacene los intentos. 
4. Haz un bucle que evalúe un número. Prueba que funcione primero. While podría ir bien
5. Modifica ese bucle para que compare nuestro número con el intento. Si se acierta usa break, si no puedes usar continue y tienes que seguir preguntando al jugador por nuevos intentos

```python
# Primero importamos la librería "random" para generar números aleatorios

import random

# Empezamos definiendo una función auxiliar "extra" que detecta si un intento ya ha sido propuesto con anterioridad

def intento_repetido(intento, intentos):                           # Comprobamos si el número 'intento' está ya dentro de la lista 'intentos'
    if intento in intentos:
        print("¡Ya has adivinado ese número y no era correcto!")   # Si está repetido, mostramos el mensaje solicitado
        return True                                                # Devolvemos True indicando que el intento es repetido
    return False                                                   # Si no está repetido, devolvemos False
    
# Ahora si, definimos la función principal del juego:

def juego_caliente_frio():
    
    # Añadimos un mensaje de bienvenida y explicamos las reglas
    
    print("🎮 Bienvenido a 'Caliente o Frío'")
    print("Reglas:")
    print("1. Piensa en un número entre 1 y 100.")
    print("2. Si el número está fuera del rango (1-100), verás 'FUERA DE LÍMITES'.")
    print("3. En el primer intento:")
    print("   - Si tu número está dentro de 10 del número secreto ➝ '¡CALIENTE!'")
    print("   - Si está fuera de ese rango ➝ '¡FRÍO!'")
    print("4. En intentos siguientes:")
    print("   - Si estás más cerca que antes ➝ '¡MÁS CALIENTE!'")
    print("   - Si estás más lejos ➝ '¡MÁS FRÍO!'")
    print("5. El juego termina cuando adivines el número.\n")

    # Genera un número aleatorio entre 1 y 100 que el jugador no podrá ver
    
    numero_secreto = random.randint(1, 100)

    # Creamos una lista vacía para guardar los intentos del jugador
    
    intentos = []

    # Creamos un bucle infinito para seguir pidiendo intentos hasta que se acierte
    
    while True:
        try:
            intento = int(input("👉 Adivina un número entre 1 y 100: "))     # Pedimos al usuario que ingrese un número
        except ValueError:
            print("⚠️ Debes escribir un número entero.")                     # Si el jugador escribe algo que no sea número, mostramos este aviso
            continue                                                         # Volvemos al inicio del bucle

        # Verificamos si el número está dentro de los límites permitidos
        
        if intento < 1 or intento > 100:
            print("FUERA DE LÍMITES")
            continue  # Volvemos a pedir número

        # Guardamos el intento en la lista de intentos
        
        intentos.append(intento)

        # Caso: el jugador adivina el número
        
        if intento == numero_secreto:
            print(f"🎉 ¡Has acertado! El número era {numero_secreto}.")
            print(f"Te ha llevado {len(intentos)} intentos.")  # Mostramos cuántos intentos usó
            break  # Salimos del bucle (fin del juego)

        # Caso: primer intento (se compara solo con el número secreto)
        
        if len(intentos) == 1:
            
            # Si la diferencia con el número secreto es menor o igual a 10 → ¡CALIENTE!
            
            if abs(numero_secreto - intento) <= 10:
                print("¡CALIENTE!")
            else:
                
                # Si está a más de 10 → ¡FRÍO!
                
                print("¡FRÍO!")
        else:
            
            # Caso: intentos siguientes, comparamos con el anterior
            # Si este intento está más cerca del secreto que el anterior → ¡MÁS CALIENTE!
            
            if abs(numero_secreto - intento) < abs(numero_secreto - intentos[-2]):
                print("¡MÁS CALIENTE!")
            else:
                
                # Si este intento está más lejos que el anterior → ¡MÁS FRÍO!"
                
                print("¡MÁS FRÍO!")

# Punto de entrada del programa: ejecuta el juego si se lanza directamente este archivo

if __name__ == "__main__":
    juego_caliente_frio()
```
