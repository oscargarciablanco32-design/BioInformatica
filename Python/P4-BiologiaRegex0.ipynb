##### 2.- Crea dos archivos. Uno almacena los accession que empiezen por 'a' y el otro el resto

```python
accs = ['ab56', 'bh84', 'hv76', 'ay93', 'ap97', 'bd72']
```

```python
# Abrimos dos archivos nuevos en modo escritura donde a_accs.txt guardará los que empiezan por a y otros_accs.txt el resto

with open('a_accs.txt', 'w') as archivo_a, open('otros_accs.txt', 'w') as archivo_otros:
    
    for acc in accs:                            #recorre cada elemento de la lista accs
        
        if acc.startswith('a'):                 #verifica si comienza con a
            archivo_a.write(acc + '\n')         #si empieza por a lo escribimos en 'a_accs.txt'
    else:
        archivo_otros.write(acc + '\n')

#comprobamos si todo ha salido bien
print("Archivos creados correctamente:")
print("- a_accs.txt contiene los accs que empiezan por 'a'")
print("- otros_accs.txt contiene el resto")
```

##### 4.- Escribe una función que cuente el porcentaje de A de una secuencia

```python
#definimos una función que recibe la cadena de ADN como parametro

def porcentaje_A(lista_dna):
    total = len(lista_dna)          #calculamos la longitud
    num_A = lista_dna.count('A')    #contamos cuantas veces aparece la letra A
    porcentaje = (num_A / total) * 100    #lo convertimos a porcentaje
    return porcentaje

resultado = porcentaje_A(lista_dna)
print(f"Porcentaje de A:", resultado)
```

##### ¿Qué pasa si no ponemos un argumento? ¿Sabrías hacer una función encapsulada?

```python
# Si la función no tiene valor por defecto y la llamas sin argumento python te devuelve un TypeError

def porcentaje_A(lista_dna):
    lista_dna = [b.upper() for b in lista_dna]
    total = len(lista_dna)
    num_A = lista_dna.count('A')
    return (num_A / total) * 100
porcentaje_A()
```

```python
#Para hacer una funcion encapsulada usamos una función interna "helper":

def porcentaje_A_encapsulada(lista_dna):
    def _calcular_porcentaje_A(seq):
        seq = [b.upper() for b in seq]
        total = len(seq)
        if total == 0:
            return 0.0
        num_A = seq.count('A')
        return (num_A / total) * 100

    return _calcular_porcentaje_A(lista_dna)

# Como la funcion interna _calcular_porcentaje_A() no puede usarse fuera de porcentaje_A_encapsulada() la convierte en una funcion encapsulada
```

##### Testea tu función original con assert

```python
# 1. Secuencia con 50% de A
seq1 = list("AATT")
assert abs(porcentaje_A_encapsulada(seq1) - 50.0) < 0.001, "Error: debería ser 50%"

# 2. Secuencia con 100% de A
seq2 = list("AAAA")
assert abs(porcentaje_A_encapsulada(seq2) - 100.0) < 0.001, "Error: debería ser 100%"

# 3. Secuencia con 0% de A
seq3 = list("CGCG")
assert abs(porcentaje_A_encapsulada(seq3) - 0.0) < 0.001, "Error: debería ser 0%"

# 4. Secuencia mixta (3 de 9 son A -> 33.33%)
seq4 = list("AAGCTAGGC")
assert abs(porcentaje_A_encapsulada(seq4) - 33.33) < 0.5, "Error: debería ser ~33.33%"

# 5. Secuencia vacía
seq5 = list("")
assert abs(porcentaje_A_encapsulada(seq5) - 0.0) < 0.001, "Error: secuencia vacía debería devolver 0%"

print("✅ Todos los tests pasaron correctamente.")
```

#### 5.- De la lista de accessions que se proporciona obtén lo siguiente:
1. Los que tengan el número 5
2. Los que tengan la letra 'd' o la 'e'
3. Los que tengan las letras 'd' y 'e' en ese orden
4. Los que tengan las letras 'd' y 'e' en ese orden con una letra cualquiera en medio
5. Los que tengan ambas letras 'd' y 'e' en cualquier orden
6. Los que empiezan por 'x' o por 'y'
7. Los que empiezan por 'x' o 'y' y terminan por 'e'
8. Los que tienen tres o más dígitos seguidos
9. Los que terminan con 'd' seguidos de 'a' o 'r' o 'p'

```python
import re
```

```python
accs = ['xkn59438', 'yhdck2', 'eihd39d9', 'chdsye847', 'hedle3455', 'xjhd53e', '45da', 'de37dp']

# Los que tienen el numero 5
con_5 = [acc for acc in accs if '5' in acc]
print("Con número 5:", con_5)

# Los que tienen la letra d o e

con_d_o_e = [acc for acc in accs if 'd' in acc or 'e' in acc]
print("Con 'd' o 'e':", con_d_o_e)

# Los que tienen las letras d y e en ese orden

de_en_orden = [acc for acc in accs if acc.find('d') != -1 and acc.find('e') > acc.find('d')]
print("'d' y 'e' en ese orden:", de_en_orden)

# Los que tienen las letras 'd' y 'e' en ese orden con cualquier letra en medio

de_con_letra = [acc for acc in accs if any(acc[i] == 'd' and i+2 < len(acc) and acc[i+2] == 'e' for i in range(len(acc)-2))]
print("'d' y 'e' con una letra en medio:", de_con_letra)

# Los que tienen ambas letras d y e en cualquier orden

con_d_y_e = [acc for acc in accs if 'd' in acc and 'e' in acc]
print("Con 'd' y 'e' en cualquier orden:", con_d_y_e)

# Los que empiezan por x o y

empiezan_xy = [acc for acc in accs if acc.startswith(('x', 'y'))]
print("Empiezan por 'x' o 'y':", empiezan_xy)

# Los que empiezan por x o y y terminan en e

empiezan_xy_y_terminan_e = [acc for acc in accs if acc.startswith(('x', 'y')) and acc.endswith('e')]
print("Empiezan por 'x' o 'y' y terminan por 'e':", empiezan_xy_y_terminan_e)

# Los que tienen tres o más digitos seguidos

tres_digitos = [acc for acc in accs if re.search(r'\d{3,}', acc)]
print("Tres o más dígitos seguidos:", tres_digitos)

# Los que terminan con d seguidos de a r o p

terminan_darp = [acc for acc in accs if acc.endswith(('da', 'dr', 'dp'))]
print("Terminan con 'da', 'dr' o 'dp':", terminan_darp)
```
